# Data Structures - List, Dictionary, Set & Conditional Statements 

### List Creation:

```python
age_list=[24, 25, 26, 27, 28]
name_list=["Akshaya", "Architha", "Meera", "Riya", "Varun"]
print(age_list)
print(name_list)
```

### List Operations / Modifications:

```python
name_list.append("Yazhini")
age_list.insert(2,30)
print(name_list)
print(age_list)
```

```python
name_list.remove("Yazhini")
age_list.pop()
print(name_list)
print(age_list)
```

```python
age_list.extend([29, 30, 26])
print(age_list)
```

```python
age_list.sort(reverse=True)
print("Descending order:", age_list)
```

```python
print("Maximum age:", max(age_list))
print("Minimum age:", min(age_list))
print("Sum of ages:", sum(age_list))
```

###  Accessing List Elements:

```python
print("First Element:", name_list[0])
```

```python
print("Last Element:", name_list[-1])
```

```python
print("Index 2 to 4:", name_list[2:5])
```

```python
print("Name list in reverse order:", name_list[::-1])
```

```python
print("Akshaya:", student_marks["Akshaya"])
```

```python
student_marks.update({"Janani": 80})
print(student_marks)
```

```python
student_marks.update({"Meera": 82})
print(student_marks)
```

```python
print("Keys:", student_marks.keys())
```

```python
print("Values:", student_marks.values())
```

```python
print("Items:", student_marks.items())
```

### Sets (Operations): 

```python
my_set={'a','e','i','o','u','a','a','i'}
print(my_set)
```

* The given values contain vowels [a, e, i, o, u]. The vowels "a" and "i" are repeated, but a set does not allow duplicate values. Therefore, each vowel appears only once in the output.

```python
 my_set[4] = 's'
```

* A set does not support indexing or changing individual elements because sets are unordered collections. Therefore, we cannot change an element using an index like my_set[4].

```python
set1 = {1, 3, 5, 7, 9}

set2 = {2, 3, 5, 8, 10}

print("Set 1:", set1)
print("Set 2:", set2)
```

```python
set3=set1.union(set2)
print("set3:", set3)
```

```python
set4=set1.intersection(set2)
print("set4:", set4)
```

### Operators & Conditional Statements : 
(IF, ELIF, ELSE) 

```python
score = int(input("Enter your score (0 to 10): "))
```

```python
if score > 7:
    print("Above Average")
elif score >= 4:
    print("Average")
else:
    print("Below Average")
```

```python
if score > 7:
    print("Above Average: Excellent performance! Keep it up.")

elif score >= 4:
    print("Average: Good effort! Keep practicing.")

else:
    print("Below Average: Need to improve your performance. Keep practicing.")
```
